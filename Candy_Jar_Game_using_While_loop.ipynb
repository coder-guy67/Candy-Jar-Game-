```python
import random
ans="y"

name=input("What is you name ")
print(f'\n')

guessctr=0
guess=0

num=random.randint(40,55)

while guess!=num and guessctr<5:
  guess=int(input("Enter your guess: "))
  guessctr+=1
  if guess>num:
    print("A little lower")
  elif guess<num:
    print("A little higher")
  else:
    print("You got your Candies")

  if guessctr==5 and guess!=num:
    print("Sorry you have run out of guesses...Good try")

  print(f'\n')
  print("That was fun! Bye "+ name +"!")
```
